```python
df["LossRatio"] = df["TotalClaims"] / df["TotalPremium"]

df["Margin"] = df["TotalPremium"] - df["TotalClaims"]

df[["LossRatio", "Margin"]].head()
```

```python
province_analysis = df.groupby("Province")[
    ["TotalPremium", "TotalClaims"]
].mean()

province_analysis
```

```python
province_analysis.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Premium and Claims by Province")

plt.ylabel("Amount")

plt.show()
```

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"])
```

```python
monthly_claims = df.groupby(
    df["TransactionDate"].dt.to_period("M")
)["TotalClaims"].mean()

monthly_claims.plot(figsize=(12,6))

plt.title("Monthly Claim Trends")

plt.show()
```

```python
df.groupby("AutoMake")["TotalClaims"].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby("AutoMake")["TotalClaims"].mean().sort_values().head(10)
```

### Observation
- Gauteng province shows higher average claim amounts.
- SUVs appear associated with higher premiums.
- Claim distribution is highly right-skewed with significant outliers.

### Key Insight
Higher vehicle values appear associated with larger claim amounts.

### Business Insight
